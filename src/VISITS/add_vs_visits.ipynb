```python
df_urg_cohort = df_urg[df_urg['idCas'].isin(switchers_1517)].copy()
df_urg_cohort['dat_ini'] = pd.to_datetime(df_urg_cohort['dat_ini'], format='%Y%m%d')
df_urg_cohort = df_urg_cohort.merge(first_lai_dates, on='idCas', how='left')
df_urg_cohort['days_to_switch'] = (df_urg_cohort['dat_ini'] - df_urg_cohort['first_lai_date']).dt.days
df_urg_cohort = df_urg_cohort[(df_urg_cohort['days_to_switch'] >= -365) & (df_urg_cohort['days_to_switch'] <= 365)]
df_urg_cohort['period'] = df_urg_cohort['days_to_switch'].apply(lambda x: 'pre' if x < 0 else 'post')

print(df_urg_cohort.shape)
print(df_urg_cohort['period'].value_counts())
```
